# Download and filter image datasets

Load MNIST, Fashion-MNIST, or CIFAR-10, keep selected targets, and cap the number of images per target.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "vision" / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

VISION_SRC = PROJECT_ROOT / "vision" / "src"
if str(VISION_SRC) not in sys.path:
    sys.path.insert(0, str(VISION_SRC))

from mimic_vision import load_vision_dataset
```

```python
DATASET = "mnist"  # "mnist", "fashion_mnist", or "cifar10"
TARGETS = [3, 8]
N_PER_TARGET = 200
SPLIT = "train"
RANDOM_STATE = 0
```

```python
dataset = load_vision_dataset(
    DATASET,
    targets=TARGETS,
    n_per_target=N_PER_TARGET,
    split=SPLIT,
    random_state=RANDOM_STATE,
)

dataset.X.shape, dataset.images.shape, dataset.y.value_counts().sort_index()
```
