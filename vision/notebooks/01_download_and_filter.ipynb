# Download, filter, and serialize image datasets

Load MNIST, Fashion-MNIST, or CIFAR-10, keep selected targets, cap the number of images per target, and save the prepared dataset with a readable filename.

```python
%load_ext autoreload
%autoreload 2
```

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "vision" / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

VISION_SRC = PROJECT_ROOT / "vision" / "src"
if str(VISION_SRC) not in sys.path:
    sys.path.insert(0, str(VISION_SRC))

from mimic_vision import load_vision_dataset, save_vision_dataset
```

```python
DATASET = "mnist"  # "mnist", "fashion_mnist", or "cifar10"
TARGETS = [3, 8]
N_PER_TARGET = 20
SPLIT = "train"
RANDOM_STATE = 0
RESIZE_SCALE = 0.5  # 0.5 halves each image side length; 28x28 becomes 14x14.
OUTPUT_DIR = PROJECT_ROOT / "vision" / "data" / "serialized"
```

```python
dataset = load_vision_dataset(
    DATASET,
    targets=TARGETS,
    n_per_target=N_PER_TARGET,
    split=SPLIT,
    random_state=RANDOM_STATE,
    resize_scale=RESIZE_SCALE,
)

summary = dataset.y.value_counts().sort_index().rename_axis("target").to_frame("count")
dataset.X.shape, dataset.images.shape, summary
```

```python
saved_path = save_vision_dataset(dataset, output_dir=OUTPUT_DIR, split=SPLIT)
saved_path.name
```
