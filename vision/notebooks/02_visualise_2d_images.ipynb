# Visualise images in 2D

Compute a 2D layout and draw small thumbnails only where they do not overlap.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
while PROJECT_ROOT != PROJECT_ROOT.parent and not (PROJECT_ROOT / "vision" / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

VISION_SRC = PROJECT_ROOT / "vision" / "src"
if str(VISION_SRC) not in sys.path:
    sys.path.insert(0, str(VISION_SRC))

from mimic_vision.datasets import load_vision_dataset
from mimic_vision.visualization import compute_2d_embedding, plot_image_embedding
```

```python
DATASET = "mnist"  # "mnist", "fashion_mnist", or "cifar10"
TARGETS = [3, 8]
N_PER_TARGET = 250
SPLIT = "train"
LAYOUT_METHOD = "pca"  # "pca" or "tsne"
RANDOM_STATE = 0
```

```python
dataset = load_vision_dataset(
    DATASET,
    targets=TARGETS,
    n_per_target=N_PER_TARGET,
    split=SPLIT,
    random_state=RANDOM_STATE,
)
coordinates = compute_2d_embedding(dataset.X, method=LAYOUT_METHOD, random_state=RANDOM_STATE)
dataset.X.shape, coordinates.shape
```

```python
fig, ax, shown = plot_image_embedding(
    coordinates,
    dataset.images,
    labels=dataset.y,
    target_names=dataset.target_names,
    title=f"{DATASET}: targets {TARGETS}",
    max_images=180,
    min_distance_px=32,
    zoom=0.55,
)
len(shown)
```
